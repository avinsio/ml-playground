### Create Numpy Arrays

```python
import numpy as np
```

**From built-in python conatiner(Lists, tuple etc)**

```python
# using List
# Convert a Python list into a NumPy array
np.array([1,2,3,4])
```

```python
# using tuple
# Convert a Python tuple into a NumPy array
np.array((1,2,3,4))
```

**Upcasting in Numpy**

* Upcasting in NumPy refers to the automatic conversion of data types when performing operations involving arrays of different types, so that no data is lost and the result has a compatible type that can hold all input values.

```python
# NumPy promotes the integers to float to use a common dtype
np.array([1,2.5,3,4])
```

**Two Dimensional numpy array**

```python
np.array([
    [1,2,3],
    [4,5,6
    ]
])
```

**minimum dimensions 2**

```python
# ndmin=2 ensures that the array has at least 2 dimensions
np.array([1,2,3,4], ndmin=2)
```

**dtype**

```python
# dtype=complex creates an array of complex numbers
np.array([1,2,3,4], dtype=complex)
```

**np.asarray**
- np.array() => Always creates a new array (i.e., it copies data). Even if the input is already a NumPy array
- np.asarray() => Just ensure object is a NumPy array

```python
a = [1,2]
# Convert the input to a NumPy array without an unnecessary copy
np.asarray(a)
```

```python
a = np.array([1,2]) # Existing arrays are not copied
a
```

```python
# Convert the input to a NumPy array without an unnecessary copy
np.asarray(a) is a
```

```python
# False means a separate array object was created
np.array(a) is a
```

*If dtype is set, array is copied only if dtype does not match:*

```python
a = np.array([1,2], dtype=np.float32)
```

```python
# A different dtype requires a new array
np.asarray(a, dtype=np.float64) is a # it create new array becoase data type is diffrenetr
```

**Read and write data from a file**

*1. Text file(.txt, .csv)*

```python
# Read numerical data from a comma-separated text file
data = np.loadtxt('data/data.txt', delimiter=',')
data
```

```python
data = data ** 2
data
```

```python
# Save the array to a text file
np.savetxt("data/output_data.txt", data, fmt="%f")
```

*2. Binary file(.npy,.npz)*

```python
# Convert a Python list into a NumPy array
arr = np.array([1,2,3,4])
np.save("data/array.npy",arr)
```

```python
# Load the saved NumPy array
arr = np.load("data/array.npy")
print(arr)
```

```python
# save multiple arrays
a =np.array([1,2,3])
b = np.array([4,5,6])
# Save multiple arrays in one .npz file
np.savez('data/arrays.npz', arr1=a, arr2=b)
```

```python
# load them back
# Load the .npz file containing multiple arrays
data = np.load("data/arrays.npz")
print(data)
```

```python
print(data["arr1"])
print(data["arr2"])
```

| File Format | Read Function | Write Function | Notes |
|---|---|---|---|
| `.txt` / `.csv` | `loadtxt()` / `genfromtxt()` | `savetxt()` | Use `delimiter` for custom separators |
| `.npy` | `load()` | `save()` | For a single NumPy array (binary format) |
| `.npz` | `load()` | `savez()` / `savez_compressed()` | For storing multiple arrays |

**Advantages of `.npy` and `.npz` over Text Files**

| Feature | `.npy` / `.npz` (Binary) | `.txt` / `.csv` (Text) |
|---|---|---|
| Efficiency | Much faster to read/write | Slower I/O |
| Storage Size | More compact | Larger file size |
| Multiple Arrays | `.npz` can store multiple arrays in one file | Not directly supported; usually requires multiple files |
| No Parsing Needed | Directly loads as a NumPy array | Requires delimiter and dtype parsing |
| Reliability | Less sensitive to formatting issues | Can be affected by extra spaces, missing values, or formatting inconsistencies |


**numpy.fromfunction**

```python
# Create values using row and column indices
np.fromfunction(lambda i,j : i ==j ,(3,3), dtype=int)
```

```python
# Calculate each value from its row and column indices
np.fromfunction(lambda i,j : i +j ,(3,3), dtype=int)
```

```python
# A new 1D array initializes from text data in a string
# Convert numbers stored in a string into a NumPy array
np.fromstring('1 2', dtype=int, sep=' ')
```

```python
x1 = [1,2,3]
x2 = [4.0,5.0,6.0]
x3 = ["a",'b','c']
```

```python
a = list(zip(x1, x2, x3))
```

**data types**

```python
my_list = [1,2,3]
arr = np.array(my_list)
print(type(arr))
print(arr)
```

```python
my_mat = [[1,2,3],[4,5,6],[7,8,9]]
mat = np.array(my_mat)
print(type(mat))
print(mat,"\n-----------")
print("dimension of the matrix: ", mat.ndim, sep='')
```

```python
#size gives the total number of elements
print("Size of this matrix: ", mat.size,sep='')

#shape gives the number of elements along each axes (dimension)
print("Shape of this matrix: ", mat.shape,sep='') 

#dtype gives the data type contained in the array
print("Data type of this matrix: ", mat.dtype,sep='') 
```

```python
my_mat = [[1.1,2,3],[4,5.2,6],[7,8.3,9]]
mat = np.array(my_mat)

print("Data type of the modified matrix: ", mat.dtype,sep='') 
#dtype gives the data type contained in the array
```
