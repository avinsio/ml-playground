**5. Supported Data Types in NumPy**
- NumPy arrays (ndarrays) are homogeneous, meaning their elements normally share the same data type.

*1. Integer Data Type*

```python
np.array([1, 2, 3])
```

```python
# dtype lets us explicitly control how the values are stored.
# int32 uses 32 bits (4 bytes) per integer.
np.array([1, 2, 3], dtype=np.int32)
```

*2. Float Data Type*

```python
np.array([1.2, 2.2, 3.2])
```

```python
# float32 stores decimal values using 32 bits per value.
np.array([1.2, 2.2, 3.2], dtype=np.float32)
```

```python
# Converting floats to integers removes the decimal part.
# For example, 1.2 becomes 1.
np.array([1.2, 2.2, 3.2], dtype=np.int32)
```

```python
# When different numeric types are mixed, NumPy promotes them
# to a common type that can represent all values.
np.array([1, 2.7]) # NumPy automatically upcasts them to the most compatible common type.
```

*3. Boolean Data Type*

```python
np.array([True, False, True], dtype=bool)
```

```python
# Boolean values are converted to integers: True → 1 and False → 0.
np.array([True, False, True], dtype=np.int32)
```

*4. String Data Type*

```python
# NumPy stores these strings using a Unicode string dtype.
np.array(["apple", "Banana"])
```

```python
# U10 means a Unicode string with a maximum length of 10 characters.
np.array(["apple", "Banana"], dtype='U10')  # Unicode strings, each up to 10 characters long
```

```python
# U3 limits each string to 3 characters, so longer strings are truncated.
np.array(["apple", "Banana"], dtype='U3')  # U3 means each string can contain up to 3 characters
```

```python
# Mixing data types
# Mixing integers and strings causes NumPy to choose a common compatible dtype.
# Here, the integers are converted to strings.
np.array([1,'2',3])   
```

```python
# This works because the string '2' can be converted to an integer.
np.array([1,'2',3], dtype=np.int32)
```

```python
# This raises an error because 'b' cannot be converted to an integer.
np.array([1,'b',3], dtype=np.int32)
```

**6. Record Arrays or Structured Arrays**
- Structured arrays in NumPy are homogeneous at the array level, meaning each element follows the same structure.
- You can define a data type and an optional name for each field (column). This makes the data easier to access, sort, and filter, similar to working with Excel, CSV files, or relational databases.

```python
# Each tuple represents one record: (name, age, marks).
# A Python list of tuples can later be converted into a structured NumPy array.
records = [
    ("aman", 22,90),
    ("ankit",15, 68),
    ("akash",18, 90),
    ("aditya",22, 80)
]
```

```python
records
```

```python
# Without a structured dtype, NumPy finds one common dtype for the entire array.
# Because names are strings, the numeric values are also converted to strings.
data = np.array(records)
data
```

```python
# A structured dtype lets each field have its own data type and name.
# This keeps name as a string and age/marks as integers.
data = np.array(records, dtype = [("name", "U15"), ("age", np.int32),("marks",np.int32)])
data
```

**You can specify `dtype` (data type) in several equivalent ways:**

- Using NumPy type strings (compact format): `[("name", "U15"), ("age", "i4"), ("marks", "i4")]`
- Using NumPy type objects: `[("name", np.unicode_), ("age", np.int32), ("marks", np.int32)]`
- Mixing Python built-in types and NumPy types: `[("name", str), ("age", int), ("marks", np.int32)]`

```python
# Access the first complete record.
data[0]
```

```python
# Access the first complete record.
print(data[0])
```

```python
# Field names let us select one column directly.
data["name"]
```

**By default, NumPy arrays (ndarrays) are homogeneous, meaning their elements share the same data type. This provides several advantages:**

- 1. Vectorized operations
- 2. Efficient memory layout (contiguous blocks)
- 3. Fast computations using C under the hood

* **NumPy** is designed to work efficiently with fixed-size elements, especially numerical data such as integers and floats. It also supports strings, booleans, dates, and custom structured data types.

* All elements in a standard NumPy array share the same data type (`dtype`).

* The data is stored in a contiguous block of memory, which improves memory efficiency and computational speed.

NumPy can also represent heterogeneous data using object arrays or structured dtypes, but object arrays generally lose many of NumPy's performance advantages.

```python
# An object array can store different Python object types in the same array.
# This provides flexibility, but usually sacrifices NumPy's performance advantages.
het = np.array([5,4.5, True, "6.5","C"])
het
```

```python
# dtype='U32' means the values are stored as Unicode strings.
# NumPy converted the mixed values to a common string representation.
het.dtype  # The mixed values are converted to a common compatible type
```

```python
het[0]
```

```python
# NumPy treats this as an array of generic Python objects.
# Object arrays generally lose the performance benefits of NumPy's vectorized operations.
# Operations such as addition and broadcasting may therefore be less efficient.
# An object array can store different Python object types in the same array.
# This provides flexibility, but usually sacrifices NumPy's performance advantages.
het = np.array([5,4.5, True, "6.5","C"], dtype=object)
het
```

```python
het.dtype
```

```python
# With dtype=object, the original Python object type is preserved.
print(type(het[0]))
```
