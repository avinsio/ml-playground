### Creating Numpy arrays and matrix Using built-in Funcations

**Arange and Linspace**

````python
import numpy as np
````

```Python 
range(start, stop, step)
np.arange(start, stop, step, dtype)
```

````python
list(range(1,10,2))
````

````python
np.arange(5,16,2, dtype=np.float32)
````

````python
np.arange(5,16,2.5)
````

````python
# Every 5th from 50 in reverse order
np.arange(50,-1,-5)
````

**np.linspace**
```python
np.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)
```

````python
np.linspace(1,5, 21)
````

````python
help(np.linspace)
````

````python
np.linspace(1,5,10, dtype=np.int32)
````

**np.logspace()**
```python
np.logspace(start, stop, num=50, endpoint=True, base=10.0, dtype=None)

````python
a = np.logspace(1,2,5)  # power(10,a)
a
````

````python
print(a[1] / a[2])
print(a[2] / a[3])
print(a[3] / a[4])
````

````python
np.linspace(1,1024, num=5, dtype=np.int32)
````

````python
np.logspace(0,10, num=5, base=2, dtype=np.int32)
````

**Numpy array and matrix creation**

````python
print("Vector of zeros:\n","-"*25)
np.zeros(5)
````

````python
print("matrix of zeros:\n","-"*25)
np.zeros((3,3))
````

````python
print("Vector of ones:\n","-"*25)
np.ones(5)
````

````python
print("Matrix of ones:\n","-"*25)
np.ones((3,3))
````

````python
np.ones((3,4), dtype=np.int32)
````

````python
print("Matrix of 5's:\n","-"*25)
5*np.ones((3,3))
````

````python
print("empty Matrix :\n","-"*25)
np.empty((4,5))
````
