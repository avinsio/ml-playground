### Rename columns

```python
import pandas as pd
```

```python
ufo = pd.read_csv("http://bit.ly/uforeports")
```

```python
ufo.head()
```

**Check column names**

```python
ufo.columns
```

**Method 1 — rename()**

```python
ufo.rename(columns={
    "Colors Reported" : "Colors_Reported",
    "Shape Reported"  : "Shape_Reported"
},inplace=True)
```

```python
ufo.head()
```

**Method 2 — Replace all column names**

```python
ufo.columns
```

```python
ufo_cols = ['City', 'Colors Reported', 'Shape Reported', 'State', 'Time']
```

```python
ufo.columns = ufo_cols
```

```python
ufo.head()
```

**Method 3 — Rename while reading data**

```python
ufo = pd.read_csv("http://bit.ly/uforeports", names=ufo_cols, header=0)
```

```python
ufo.head()
```

**Method 4 — Transform all column names using str method**

```python
ufo.columns
```

```python
ufo.columns = ufo.columns.str.replace(" ","_")
```

```python
ufo.head()
```
