### Select rows & columns

```python
import pandas as pd
```

```python
ufo = pd.read_csv("http://bit.ly/uforeports")
```

```python
ufo.head()
```


**Loc = Select rows and columns by label**
- Label-based selection; label slices are inclusive on both ends.

**Single rows**

```python
ufo.loc[0,:]
```

**Multiple rows**

```python
ufo.loc[[0,1,2], :]
```

```python
ufo.loc[0:2, :]
```

```python
ufo.loc[0:2]
```

**Column selection**

```python
ufo.loc[:,["City","State"]]
```

```python
ufo.loc[:,"City":"State"]
```

**Select rows and column**

```python
ufo.loc[0:2,"City":"State"]
```

```python
ufo.head(3).drop('Time', axis=1)
```

```python
ufo[ufo.City=="Oakland"]
```

```python
ufo.loc[ufo.City=="Oakland","State"]
```

```python
ufo[ufo.City=="Oakland"].State
```

**Iloc - select rows and columns by interger position**

```python
ufo.iloc[:,[0,3]]
```

```python
ufo.columns
```

```python
list(range(0,4))
```

```python
ufo.iloc[:,0:4]
```

```python
ufo.iloc[0:3,:]
```
