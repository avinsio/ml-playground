### Axis parameter

```python
import pandas as pd
```

```python
drinks = pd.read_csv("http://bit.ly/drinksbycountry")
```

```python
drinks.head()
```

**Axis 1 = columns**

```python
drinks.drop('continent', axis=1).head()
```

**Axis 0 - rows**

```python
drinks.drop(2, axis=0).head()
```

```python
drinks.mean(axis=0, numeric_only=True)
```

```python
drinks.mean(axis=1, numeric_only=True)
```

```python
drinks.mean(axis=1, numeric_only=True).shape
```

```python
drinks.mean(axis=0, numeric_only=True).shape
```

#### Another way to do

**axis = index**

```python
drinks.mean(axis='index', numeric_only=True)
```

**axis ="columns"**

```python
drinks.mean(axis='columns', numeric_only=True)
```
