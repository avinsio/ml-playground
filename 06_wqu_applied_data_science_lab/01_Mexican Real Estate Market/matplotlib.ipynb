### Matplotlib

**Our dataset**

```python
import numpy as np
np.random.seed(42)
data ={
    "category": ["Chairs","Tables", "Curtains","Flower Pots"],
    "sales" : [150, 230, 180, 290],
    "profit" : [45, 82, 55 , 95]
}
data
```

**1. Implicit style (pyplot)**

```python
# bar plot
import matplotlib.pyplot as plt
plt.figure(figsize=(9, 3.5))
plt.bar(data["category"], data["sales"], color="steelblue")
plt.xlabel('Category')
plt.ylabel("Sales")
plt.title("Sales by category")
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

**2.Explicit Approach**

```python
fig, ax = plt.subplots(figsize=(9,3.5))
ax.bar(data["category"], data['sales'], color="steelblue")
ax.set_xlabel('Category')
ax.set_ylabel('Sales')
ax.set_title("Sales by category")
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

Multiple plot

```python
fig, ax = plt.subplots( 1, 2, figsize=(9,3.))

#left plot
ax[0].bar(data["category"], data['sales'], color="steelblue")
ax[0].set_title("Sales by category")
ax[0].set_ylabel('Sales')

# right plot
ax[1].bar(data["category"], data['profit'], color="green")
ax[1].set_title('profit by category')
ax[1].set_ylabel('profit')

plt.tight_layout()
plt.show()
```

**Pandas plot**

```python
import pandas as pd
df = pd.DataFrame(data)
df
```

```python
fig, ax = plt.subplots(figsize=(9,3.5))

df.plot(
    x ="category", y='sales', kind ='bar',
    ax=ax, title ="Sales by category",
    ylabel ="Sales", color='steelblue', legend=False
)
plt.tight_layout()
plt.show()
```

**Seaborn**

```python
import seaborn as sns

fig, ax = plt.subplots(figsize=(9,3.5))
sns.scatterplot(data=df, x="sales",y="profit",s=100, color="steelblue", ax=ax)
ax.set_title("Profit vs Sales")
ax.set_xlabel("Sales")
ax.set_ylabel("Profit")
plt.tight_layout()
plt.show()
```

Regression plot

```python
fig, ax = plt.subplots(figsize=(9,3.5))
sns.regplot(data=df, x="sales",y="profit", color="steelblue", ax=ax)
ax.set_title("Profit vs Sales(with Linear regression)")
ax.set_xlabel("Sales")
ax.set_ylabel("Profit")
plt.tight_layout()
plt.show()
```
