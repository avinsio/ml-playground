```python
#  Drop missing values
(
    pd.read_csv(path_1)
    .dropna()
)   
```

```python
df1 = (
    pd.read_csv(path_1)
    .dropna()
    .assign(
        price_usd = lambda x : x["price_usd"]
        .str.replace("$","",regex=False)
        .str.replace(",","",regex=False)
        .astype(float)
    )
)
```

```python
path_2 = "./data/mexico-real-estate-2.csv"
path_3 = "./data/mexico-real-estate-3.csv"
```

```python
df2 = (
    pd.read_csv(path_2)
    .dropna()
    .assign(
        price_usd = lambda x : x['price_mxn'] /19
    )
    .drop("price_mxn", axis=1)
)
print(f"df2 type: {type(df2)}")
print(f"df2 shape: {df2.shape}")
df2.head()
```

```python
df3 = (
    pd.read_csv(path_3)
    .assign(
        lat = lambda x: x["lat-lon"]
        .str.split(",",expand=True)[0]
        .astype(float),
        lon = lambda x : x['lat-lon']
        .str.split(",", expand=True)[1]
        .astype(float),
        state = lambda x : x["place_with_parent_names"]
        .str.split("|",expand=True)[2]
        
    )
    .drop(["place_with_parent_names","lat-lon"], axis=1)
)
print(f"df3 type: {type(df3)}")
print(f"df3 shape: {df3.shape}")
df3.head()
```

```python
# Check if all column sets are identical
columns_match = set(df1) == set(df2) == set(df3)

if columns_match:
    print("All DataFrames have the same columns.")
else:
    print("Column mismatch detected!")
    # Optional: See the difference
    print("df1 vs df2 diff:", set(df1.columns) ^ set(df2.columns))
    print("df2 vs df3 diff:", set(df2.columns) ^ set(df3.columns))
```

```python
# Stack df1, df2, and df3 vertically
df = pd.concat([df1, df2, df3], ignore_index=True)
df.shape
```

```python
# Save your cleaned combined dataset
df.to_csv("./data/mexico-real-estate-combined-clean.csv", index=False)
```

**Descriptive Statistics**

```python
# Summarize the numeric columns
summary = df[["price_usd", "area_m2"]].describe()

# Format for readability
pd.options.display.float_format = '{:,.2f}'.format

summary
```

Reading `.describe()` 

price_usd

* Count: Should be ~1,700+; lower means some prices are still NaN.
* Mean > Median: Indicates right skew; expensive properties pull the mean upward.
* Std: High variation means price depends on factors like location, area, and property type.
* Min/Max: Check for suspicious values or genuine luxury properties.

area_m2

* Also likely right-skewed: most properties are smaller, with a few very large ones.
* IQR = Q3 − Q1: Shows the middle 50% of properties.

For our research question

Since we're comparing prices across states, median is the better primary measure because it is less affected by extreme luxury properties.

What `.describe()` cannot show

* Distribution shape or multiple peaks
* Where outliers occur
* Price–area relationship
* Geographic patterns

Visualization will reveal these patterns more clearly.
