# Gradient descent

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1.0,2.0])
y = np.array([200.0,400.0])
```

```python
# cost funcation
def compute_cost(X, y, w, b):
    cost = 0
    m = len(X)

    for i in range(m):
        error = (w * X[i] + b) - y[i]
        se = error ** 2
        cost += se / (2 * m)

    return cost
```

```python
# compute gradient
def comput_gradient(X,y,w,b):
  m = len(X)
  dj_dw = 0
  dj_db =0
  for i in range(m):
    pred = w * X[i] + b
    error = pred - y[i]
    dj_dw_i = error * X[i]
    dj_db_i = error
    dj_dw += dj_dw_i
    dj_db += dj_db_i
  dj_dw = dj_dw / m
  dj_db = dj_db /m
  return dj_dw , dj_db
```

```python
# gradient descent
def gradient_descent(x, y, w, b, alpha, num_iters):

    for i in range(num_iters):

        dj_dw, dj_db = comput_gradient(x, y, w, b)

        w = w - alpha * dj_dw
        b = b - alpha * dj_db

    return w, b
```

```python
w_init = 0
b_init = 0

alpha = 0.01
num_iters = 1000

w_final, b_final = gradient_descent(
    X, y, w_init, b_init, alpha, num_iters
)

print(w_final)
print(b_final)
```
