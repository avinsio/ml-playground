In a DataFrame:

- Index → The row labels of the DataFrame. By default, pandas assigns `0, 1, 2, ...` as the index.
- Columns → The labels for each vertical series in the DataFrame.
- Labels → Identifiers for both rows (index) and columns. They are commonly used with `.loc`.

| Component | Description | Example |
|---|---|---|
| Index | Row labels | `0, 1, 2` or custom |
| Columns | Column labels | `"Name"`, `"Age"`, `"City"` |
| Labels | Identifiers for rows/columns | Used in `.loc` |

**Selecting/ indexing rows**
- Label-based `loc` method
- index(numeric) `iloc` method

```python
df.head()
```

```python
# selecting row ith label - C or index 2 ==> subsetting
print(df.loc['C'])
print("-"*100)
print(df.iloc[2])
```

```python
# Select multiple rows with label - "A" and "C" ==> Subsetting
print(df.loc[["A","C"]])

print("-"*100)

print(df.iloc[[0,2]])
```

```python
# Slicing for multiple rows - first 3 rows ==> slicing
print(df.loc["A":"C"])
print("-"*100)
print(df.iloc[0:2])
```

**Subsetting Dataframe**
- subsetting refrers to filtering down the DataFrame. Selecting a subset of rows, columns, or both

```python
df
```

```python
# Elemey at row "B" and column "Y" is
print(df.loc["B", "Y"])
print("-"*100)
print(df.iloc[1,1])
```

```python
# subset comparising of rows B and C and column w and y
print(df.loc[["B","C"], ["W", "Y"]])
print("-"*100)
print(df.iloc[[1,2], [0,1]])
```

```python
# Creating a dataframe
# creating a dataframe
matrix_data = np.matrix('22,160,70;42,170,80;30,175,90;35,160,70;25,140,70')
row_labels = ['A','B','C','D','E']
column_headings = ['Age', 'Height', 'Weight'] # 'Height(Cm)', 'Weight(Kg)
```

```python
matrix_data
```

```python
df = pd.DataFrame(data =matrix_data, index=row_labels, columns=column_headings)
print("\nA new DataFrame\n","-"*25, sep="")
print(df)
```

```python
print("\nRows with height > 158cm\n", sep='')
print(df[df["Height"]> 158])
```

```python
print("\nRows with height > 158 and weight > 70\n", sep="")
print("-"*50)
print(df[((df["Height"] > 158) & (df["Weight"] > 70))])
```
