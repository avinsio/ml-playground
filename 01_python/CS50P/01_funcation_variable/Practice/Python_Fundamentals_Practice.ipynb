**Q12. Create a **Student Introduction Program**.**

Ask the user for:
- Name
- Age
- College
- Course
- Favorite technology

Then display a clean introduction containing all five values.

### Requirements
- Use `input()`
- Store every input in a variable
- Use an f-string at least once
- Use `print()` multiple times
- Use `sep` or `end` at least once
- Include a quoted sentence such as: `"I love learning!"`

Make the output look professional and readable.

````python
name = input("What's your name? ")
age = input("What's your age? ")
college = input("What's your college name? ")
tech = input("What's your favorite technology? ")

print("My name is", name)
print(f"I am {age} years old.")
print("I study in", college, sep=" ", end=".\n")
print("My favorite technology is", tech, end=".")
````
