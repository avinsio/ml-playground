### Itroducation to Funation 
A funcation is a reusable block of code that performs a specific task

Funcations help make your code
- midular
- Reusable
- Easier to read and maintain

#### Why Use funcations?
- To avoid repetition
- To break complex problems into smaller, mangeable chunks
- To make code reusable and organized

**Problem: You need to write a Industry grade function which takes a string as input, should preprocess it by kepping only alphanumeric english and hindi characters**

````python
import re
import logging

# Configure logging
logging.basicConfig(level=logging.INFO, format='[%(levelname)s] %(message)s')


def sanitize_input(user_input: str) -> str:
    """
    Cleans user input by removing unwanted special characters
    while preserving basic punctuation and Unicode characters.

    Args:
        user_input (str): Raw user input string

    Returns:
        str: Sanitized string
    """
    # Step 1: Keep only English, Hindi characters, and spaces
    pattern = r"[^A-Za-z0-9\u0900-\u097F]"
    cleaned = re.sub(pattern, ' ', user_input, flags=re.UNICODE)

    # Step 2: Replace multiple spaces with a single space
    normalized = re.sub(r'\s+', ' ', cleaned)
    
    return normalized.strip()
````

````python
sanitize_input("नमस्कार !! 🙏 Thank you for learning with us! -> Team 🚀 Decode-AI; ")
````

#### Funation Vs Method

````python
# Funcation
def greet(name):
    print(f"hello {name}")
````

````python
# Method
class Greeter:
    def greet(self, name):
        print(f"Hello {name}")
g = Greeter()
g.greet("Avinash")
````

````python
# using return in a funcation
def add_num(num1, num2):
    return num1 + num2

add_num(12,5)
````

````python
sum_2_num = add_num(12,5)
print(str(sum_2_num))
````

````python
# check whether a number is prime or not
def is_prime(num):
    for idx in range(2, num):
        if num % idx == 0:
            return False
    return True
````

````python
if is_prime(15):
    print("Prime")
else:
    print("Not prime")
````
