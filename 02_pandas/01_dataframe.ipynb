# 1. Pandas DataFrame

A **DataFrame** is a way to store data in **rows and columns**, like a table in Excel.

Each row usually represents one record, and each column represents a piece of information about it.

### Main Features

- **Two-dimensional:** It has rows and columns.
- **Labeled:** Each row has an index label, and each column has a name.
- **Different data types:** One column can contain numbers, while another contains text or dates.
- **Mutable:** We can change values and add or remove rows and columns.

### Example

| Index | Name  | Age | City   |
|-------|-------|-----|--------|
| 0     | Aman  | 25  | Delhi  |
| 1     | Ankit | 30  | Patna  |
| 2     | Ayush | 35  | Kanpur |

Here:

- There are **3 rows and 3 data columns**.
- Each row contains information about one person.
- `Name`, `Age`, and `City` are the column names.
- `0`, `1`, and `2` are the row index labels.
- The index is separate from the data columns.

We use DataFrames to explore, clean, and analyze data.

# 3. DataFrame Internals: Why Are Pandas DataFrames Powerful?

Pandas makes data easy to work with while using efficient tools behind the scenes.

## 1. NumPy Arrays Help Make Pandas Fast

- Pandas uses **NumPy arrays** to store many types of data. Some columns use other storage types.
- A NumPy array stores values of the **same data type**, such as integers or floats.
- These values are often stored together in memory, which helps the computer process them efficiently.
- Many operations use **vectorization**: we perform an operation on a whole array or column instead of writing a Python loop.
- Much of this work runs in compiled code, which is usually faster than looping through values in Python.

**Example:**

```python
df["Age"] = df["Age"] + 1

````python
data = {
    'Name': ['Amit', 'Anand', 'Akash'],
    'Age': [25, 28, 30],
    'City': ['Mumbai', 'Delhi', 'Patna']
}
df = pd.DataFrame(data)
df
````

````python
type(df["Name"])
````

````python
type(df['Age'].values)
````

````python
df["Age"] + df["Age"]
````

Why use NumPy under the hood?
Speed – Vectorized operations are much faster than Python loops.
Memory efficiency – NumPy arrays store data more compactly than Python lists.
Interoperability – Easy to use NumPy’s mathematical and statistical functions directly on DataFrames.

## 2. Column-Based Storage

Pandas organizes data by columns. Each column contains values for one feature, such as `Age`, `Name`, or `Salary`.

**Why is this useful?**

- **Efficient column operations:** Pandas can work with the columns we need without processing every column.
- **Useful for analysis:** Tasks like finding the average salary or total sales usually work on whole columns.
- **Efficient storage:** Keeping values of the same type together can reduce memory use. Some types, such as categorical data, can save additional space.

**Example:**

```python
df["Salary"].mean()

````python
type(df['Name'])
````

## 3. BlockManager

- Pandas uses an internal system called **BlockManager** to manage a DataFrame’s data.
- It can group columns with the **same data type** into blocks.
- For example, two `float64` columns may share one block, while an integer column uses another.
- This helps Pandas perform some operations efficiently while supporting different data types in the same DataFrame.

**Note:** Columns with the same data type are not always stored in a single block.

## 4. Label-Aware Indexing

- A DataFrame has **row labels** (its index) and **column labels** (its column names).
- These labels help us select data without remembering its position.
- Pandas also uses labels to **match data automatically** during operations, even when it appears in a different order.

**Example:**

```python
df.loc["Aman", "Age"]

## 5. C and Cython Acceleration

- Many operations in Pandas use **compiled code**, including code written in C or Cython.
- **Cython** helps turn Python-like code into compiled code that can run faster.
- This helps Pandas process large datasets more quickly than typical Python loops.

**Note:** Speed still depends on the operation, data types, dataset size, and available memory. Not every Pandas operation is equally fast.

## 6. Useful Data Information and Built-in Methods

- Pandas keeps information about our data, such as column names, row labels, and data types.
- It also supports missing values and categorical data.
- It provides built-in methods for common analysis tasks:
  - **Reshaping:** Change the layout of a table using `pivot()` or `melt()`.
  - **Grouping:** Summarize data by category using `groupby()`.
  - **Combining:** Bring tables together using `merge()` or `join()`.
  - **Working with dates and times:** Extract dates, calculate time differences, and summarize data by day or month.

These tools help us analyze data without writing everything from scratch.
