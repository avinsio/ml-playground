### Read the tabular data in Pandas

```python
import pandas as pd
```

```python
orders = pd.read_table('http://bit.ly/chiporders')
```

```python
# first five rows
orders.head()
```

```python
user_cols = ["user_id","age","gender","occupation","zip_code"]
users = pd.read_table("http://bit.ly/movieusers", sep="|", header=None, names=user_cols)
```

```python
users.head()
```

```python
orders = pd.read_table('http://bit.ly/chiporders', skiprows=[1,3,4])
orders.head()
```
