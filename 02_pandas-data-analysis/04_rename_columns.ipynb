### How do i rename columns in a Pandas DataFrame?

```python
import pandas as pd
```

```python
ufo = pd.read_csv("http://bit.ly/uforeports")
ufo.head()
```

```python
# check column name
ufo.columns
```

**Rename column name**

1. rename method

```python
ufo.rename(columns= {
    'Colors Reported' : 'Colors_Reported',
    'Shape Reported' : 'Shape_Reported'
}, inplace=True)
ufo.columns
```

2. using python list

```python
ufo_cols = ['City', 'Colors Reported', 'Shape Reported', 'State', 'Time']
ufo.columns = ufo_cols
```

```python
ufo.head()
```

3. Rename column while loading data

```python
ufo = pd.read_csv("http://bit.ly/uforeports", names = ufo_cols, header =0)
ufo.head()
```

**Bonus tips**

```python
# Str method
ufo.columns = ufo.columns.str.lower().str.replace(" ","_")
```
