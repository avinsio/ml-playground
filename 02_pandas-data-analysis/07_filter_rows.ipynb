### Filter rows

```python
import pandas as pd
```

```python
movies = pd.read_csv("http://bit.ly/imdbratings")
movies.head()
```

```python
movies.shape
```

```python
type(True)
```

```python
boolean = []
for length in movies.duration:
    if length >=200:
        boolean.append(True)
    else:
        boolean.append(False)
```

```python
boolean[:5]
```

```python
len(boolean)
```

```python
is_long = pd.Series(boolean)
is_long.head()
```

```python
is_long = movies.duration >=200
is_long
```

```python
movies[is_long].head()
```
