### How do i Select a Pandas Series from a dataframe? 

```python
import pandas as pd
```

```python
ufo = pd.read_table("http://bit.ly/uforeports", sep =",")
ufo.head()
```

```python
ufo = pd.read_csv("http://bit.ly/uforeports")
ufo.head()
```

```python
type(ufo)
```

**Ways to Select Series**

1. Using bracket notation

```python
ufo["City"]
```

```python
type(ufo["City"])
```

2. Using dot notation

```python
ufo.City
```

```python
ufo.Colors Reported   # here dot notation will not work
```
