# Python Lists

A **list** is an **ordered, mutable collection** of Python objects.

## Features of a List

1. **Ordered** — Each item has a position. In `[10, 5.6, 4, 2]`, `10` is at index `0`.
2. **Mutable** — You can add, remove, or replace items without creating a new list.
3. **Allows duplicates** — The same value can appear more than once, as in `[10, 10, 4]`.
4. **Heterogeneous** — A list can hold items of different types, as in `[10, 5.6, "hello"]`.

## A List Is Both a Container and a Sequence

- A **container** holds other objects.
- A **sequence** is an ordered collection that supports indexing and slicing. For example, `data[0]` gets the first item, and `data[1:3]` gets a new list containing the items at indices `1` and `2`.

## How Is a List Stored in Memory?

```python
data = [10, 5.6, 4, 2]
```

This list has four items. `10`, `4`, and `2` are integers, while `5.6` is a float.

In **CPython**, the name `data` refers to a list object. The list object stores its length and points to an internal array of references. Each slot in that array holds a reference to a Python object, not the value itself. The slots are next to each other in memory, but the objects they refer to don't have to be.

```mermaid
flowchart TD
    D["Name: data"] --> L["List object: length = 4"]
    L --> A["Internal array of references"]
    A --- S0["Slot 0"]
    A --- S1["Slot 1"]
    A --- S2["Slot 2"]
    A --- S3["Slot 3"]
    S0 --> O0["int object: 10"]
    S1 --> O1["float object: 5.6"]
    S2 --> O2["int object: 4"]
    S3 --> O3["int object: 2"]
```

## What Happens When You Replace an Item?

```python
data[1]       # Returns 5.6
data[1] = 99  # Changes slot 1 to refer to the integer object 99
print(data)   # [10, 99, 4, 2]
```

The assignment changes the reference in slot `1`. The list is still the **same list object**, and the float object `5.6` itself was not modified.

**Mental model:** The name refers to the list. The list's slots refer to the items.
