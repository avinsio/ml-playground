###  Univariate Analysis

Exploring one variable at a time to understand its distribution, identify patterns, and detect potential data quality issues.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../datasets/titanic.csv")
```

```python
df.head()
```

```python
df.dtypes
```

#### 1. Categorical Variables

* Categorical variables contain a fixed set of categories or labels. The goal is to understand how frequently each category appears.

**a. Countplot**

* A count plot displays the number of observations in each category. It provides a quick overview of the distribution of categorical data.

```python
sns.countplot(x="Pclass",
              data=df);
```

**Frequency Count**

* value_counts() returns the number of occurrences of each category.

```python
df["Pclass"].value_counts()
```

**Bar Chart**

* A bar chart represents category frequencies using bars. It provides the same information as a count plot but offers more customization.

```python
df["Pclass"].value_counts().plot(kind='bar');
```

**b. Pie Chart**
* A pie chart shows the proportion of each category as a percentage of the whole dataset.

```python
df["Pclass"].value_counts().plot(kind="pie", autopct="%.2f");
```

#### 2. Numerical Data
* Numerical variables contain measurable values. The goal is to understand their distribution, central tendency, spread, and potential outliers.

**a. Histogram**
* A histogram groups numerical values into intervals (bins) and shows how frequently values fall within each interval.

```python
plt.hist(df["Age"], bins=50);
```

**b. Histogram with KDE**
* The Kernel Density Estimate (KDE) provides a smooth estimate of the data distribution, making the overall shape easier to interpret.

```python
sns.histplot(x="Age",
             data=df,
             kde=True,
             stat="density");
```

**c. Boxplot**
* A box plot summarizes the distribution using the median, quartiles, and potential outliers.

```python
sns.boxplot(x="Age",
            data=df);
```

```python
df["Age"].describe()
```

**Skewness**

Skewness measures the asymmetry of a distribution.

- Positive skew → longer right tail
- Negative skew → longer left tail
- Around 0 → approximately symmetric

```python
df["Age"].skew()
```
