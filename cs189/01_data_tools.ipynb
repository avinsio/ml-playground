```python
import pandas as pd
```

#### Series
- A Series is a one-dimensional labeled array.
- `values`: The actual data stores in the series
- `index`: The label associated with each data point

```python
welcome_series = pd.Series(["Welcome", "to", "CS 189"])
welcome_series
```

```python
# values
welcome_series.values
```

```python
# index
welcome_series.index
```

##### DataFrame

**Creating a Dataframe**

```python
data = {
    'Landmark': ['Sather Gate', 'Campanile', 'Doe Library', 'Memorial Glade', 'Sproul Plaza'],
    'Type': ['Gate', 'Tower', 'Library', 'Open Space', 'Plaza'],
    'Height': [30, 307, 80, 0, 0],
    'Year Built': [1910, 1914, 1911, None, 1962]
}

df = pd.DataFrame(data)
df
```

##### Untility Funcations for DataFrame

```python
# first few rows
df.head() # default is 5
```

```python
# Last few rows
df.tail()  
```

```python
# shape
df.shape
```

```python
# size
df.size
```

```python
# random selection of rows from a dataframe
df.sample(n= 5)
```

```python
# randomly sample frac of rows
df.sample(frac= 0.1)
```

```python
# Allow the same row to appear multiple times
df.sample(n= 3, replace=True)
```

```python
# Using random_state for reproducibility
df.sample(n= 3, random_state=42)
```

```python
# Values_Counts()
type_counts = df["Type"].value_counts()
type_counts
```

```python
type_counts = df["Type"].unique()
type_counts
```
