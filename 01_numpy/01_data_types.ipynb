## 6. Record array or structured array
- Structure arrays in Numpy are homogeneous at the array level (each element has the same layout)
- we can specify a type and optionally, a name on a per-column basis. This makes sorting and filtering even more powerful, and it can feel similar to work in Excel, CSVs or realtional datbases

```python
records = [
    ("Aman", 22, 90),
    ("Ankit", 15, 60),
    ("Akash", 18, 90),
    ("Aditya", 22, 80)
]
records
```

```python
np.array(records).dtype
```

```python
data = np.array(records, dtype=[("name", "U15"), ("age", np.int32), ("marks", np.int32)])
data
```

```python
data.dtype
```

**You can specify the dtype (data type) in several equivalent ways.**

- Using NumPy type strings (compact format) - [("name", "U15"), ("age", "i4"), ("marks", "i4")]
- Using NumPy type objects (recommended) - [("name", np.unicode_), ("age", np.int32), ("marks", np.int32)]
- Mixing Python built-in types and NumPy types - [("name", str), ("age", int), ("marks", np.int32)]

```python
print(data[0])
```

```python
print(data['name'])
```

**By default, NumPy arrays (ndarray) are homogeneous, meaning all elements must be of the same data type.
Homogeneous arrays allow:**

- *Vectorized operations*
    - Efficient memory layout (contiguous blocks)
    - Fast computations using C under the hood
    - NumPy is designed for arrays of fixed-size elements — typically numerical data types (integers, floats), but also supports strings, booleans, dates, and custom structured types.

- All elements in a NumPy array share the same data type (dtype).
- The data is stored in a contiguous block of memory for efficiency.

It can support heterogeneous types using object arrays or structured dtypes, but these lose most performance benefits.

```python
het = np.array([5, 4.5, True, 6.5, "C"])
het
```

```python
het.dtype
```

```python
het[0]
```

```python
het = np.array([5, 4.5, True, 6.5, "C"], dtype='object')
het
```

```python
het.dtype
```

```python
het[0]
```
