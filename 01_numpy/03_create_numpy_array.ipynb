**np.asarray**
- np.array() => Always creates a new array (i.e., it copies data). Even if the input is already a NumPy array
- np.asarray() => Just ensure object is a NumPy array

```python
a = [1,2]
np.asarray(a)
```

```python
a = np.array([1,2])
a
```

```python
np.asarray(a)
```

```python
np.asarray(a) is a
```

```python
np.array(a) is a
```

```python
a = np.array([1,2], dtype=np.float32)
```

```python
np.asarray(a, dtype=np.float64) is a
```

**Read and write data from a file**

1. Text files(.txt,.csv)

```python
data =np.loadtxt("data/data.txt", delimiter=',')
data
```

```python
data = data **2
data
```

**Binary Files(.npy, .npz)**

```python
arr = np.array([1,2,3,4])
np.save("data/array.npy", arr)
```

```python
arr = np.load("data/array.npy")
arr
```

```python
# Save multiple arrays
a = np.array([1,2,3])
b = np.array([4,5,6])
np.savez("data/array.npz", arr1=a, arr2= b)
```

```python
# load them back
data = np.load("data/array.npz")
print(data["arr1"])
print(data["arr2"])
```

##### NumPy File Formats

| **File Format** | **Read Function** | **Write Function** | **Notes** |
|---|---|---|---|
| `.txt` / `.csv` | `loadtxt()` / `genfromtxt()` | `savetxt()` | Use a delimiter to separate values. |
| `.npy` | `load()` | `save()` | Stores a single array in binary format. |
| `.npz` | `load()` | `savez()` / `savez_compressed()` | Stores multiple arrays in one file. |

#####  Advantages of `.npy` and `.npz` Over Text Files

| **Feature** | **`.npy` / `.npz` (Binary)** | **`.txt` / `.csv` (Text)** |
|---|---|---|
| **Speed** | Faster to read and write. | Generally slower to read and write. |
| **File Size** | Usually smaller. | Usually larger. |
| **Multiple Arrays** | `.npz` can store multiple arrays in one file. | Usually requires separate files. |
| **Loading Data** | Preserves array data types and shapes. | May require parsing values and specifying data types. |
| **Formatting** | Preserves array structure. | Can be affected by formatting differences, extra spaces, or missing values. |

**Data type**

```python
my_list = [1,2,3]
import numpy as np
arr = np.array(my_list)
print("Type/Class of this object:",type(arr))
print("Here is the vector\n--------------------\n",arr)
```

```python
my_mat = [[1,2,3],[4,5,6],[7,8,9]]
mat = np.array(my_mat)

print("Type/Class of this object:",type(mat))
print("Here is the matrix\n----------\n",mat,"\n----------")

#ndim gives the dimensison, 2 for a matrix, 1 for a vector
print("Dimension of this matrix: ",mat.ndim,sep='')
```

```python
#size gives the total number of elements
print("Size of this matrix: ", mat.size,sep='')

#shape gives the number of elements along each axes (dimension)
print("Shape of this matrix: ", mat.shape,sep='') 

#dtype gives the data type contained in the array
print("Data type of this matrix: ", mat.dtype,sep='') 
```

```python
my_mat = [[1.1,2,3],[4,5.2,6],[7,8.3,9]]
mat = np.array(my_mat)

print("Data type of the modified matrix: ", mat.dtype,sep='') 
#dtype gives the data type contained in the array
```
