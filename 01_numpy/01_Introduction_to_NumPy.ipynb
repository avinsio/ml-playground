## Introduction to NumPy

### Why Learn NumPy?

- **Useful for AI and machine learning:** NumPy is commonly used when working with numerical data.
- **Important for data science:** It helps us perform calculations on large amounts of data efficiently.
- **Build algorithms from scratch:** NumPy is useful when implementing ideas such as linear regression, gradient descent, and activation functions.

#### Introduction to NumPy

- **NumPy** stands for **Numerical Python**. It is a Python library used for numerical and scientific computing.
- Its main data structure is the **ndarray**, which stores data in one or more dimensions.
- NumPy provides built-in functions to perform mathematical operations on arrays quickly and efficiently.
- It is widely used in data science, machine learning, scientific computing, and image processing.

#### Why Do We Use NumPy?

##### 1. Working with Numerical Data
NumPy arrays are useful for storing and processing numerical data. Many scientific and machine learning tools can work with NumPy arrays.

##### 2. Faster Calculations
NumPy uses optimized code written in languages such as C to perform many operations efficiently. For large numerical tasks, these operations are often faster than writing equivalent Python loops.

NumPy supports operations such as:
- Addition and multiplication of arrays
- Dot products and matrix multiplication
- **Broadcasting** — applying operations to arrays of compatible shapes without manually repeating values

##### 3. Arrays with Multiple Dimensions
NumPy supports arrays of different dimensions:
- **1D array:** similar to a list of values
- **2D array:** similar to a table or matrix
- **3D and higher-dimensional arrays:** useful for data such as images and batches of images

##### 4. Building Machine Learning Algorithms
NumPy helps us understand and implement the mathematics behind machine learning algorithms, including:
- Linear regression
- Gradient descent
- Activation functions
- Parts of neural-network training

##### NumPy vs Pandas

| Feature | NumPy | Pandas |
|---|---|---|
| Main purpose | Numerical calculations | Working with structured, tabular data |
| Main data structures | `ndarray` | `Series` and `DataFrame` |
| Labels | Usually accessed by position | Supports row and column labels |
| Common uses | Array operations and mathematical calculations | Data cleaning, analysis, and preprocessing |

**In simple terms:** Use Pandas to explore and prepare tabular data, and use NumPy for efficient numerical operations. They often work together.

#### Why Are NumPy Arrays Often More Memory-Efficient Than Python Lists?

##### Python Lists
- A Python list stores references to Python objects.
- Each number is a Python object with additional information, which creates memory overhead.
- For large collections of numbers, this overhead can use more memory.

##### NumPy Arrays
- A NumPy array usually stores elements of the same data type in a compact block of memory.
- It avoids much of the per-number object overhead found in Python lists.
- Its layout helps the computer perform numerical operations efficiently.
- NumPy can apply many operations to whole arrays at once. This is called **vectorization**, and it often avoids the need for explicit Python loops.

##### Key Takeaway
For large collections of numerical values, NumPy arrays are often more memory-efficient and faster to process than Python lists. The exact difference depends on the data type, operation, and size of the data.
